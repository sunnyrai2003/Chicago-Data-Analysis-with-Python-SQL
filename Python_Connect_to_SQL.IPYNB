```python
import pandas as pd
import sqlite3


conn = sqlite3.connect("My_Project.db")


 #read csv file
df1 = pd.read_csv("C:\\Users\\Sunny Rai\\Downloads\\ChicagoCrimeData.csv")
df2 = pd.read_csv("C:\\Users\\Sunny Rai\\Downloads\\ChicagoPublicSchools.csv")
df3 = pd.read_csv("C:\\Users\\Sunny Rai\\Downloads\\ChicagoCensusData.csv")


df1.to_sql("crime_data", conn, if_exists="replace", index=False)
df2.to_sql("school_data", conn, if_exists="replace", index=False)
df3.to_sql("census_data", conn, if_exists="replace", index=False)

print("All 3 CSV files uploaded successfully!")
```

```python
query = """
SELECT *
FROM crime_data
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result.head(3))
```

```python
query = """
SELECT count(*) as total_rows
FROM crime_data;
"""

result = pd.read_sql_query(query, conn)

print(result.head(3))
```

```python
query = """ select count(*) as total_rows from school_data;"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
query = """ select count(*) as total_rows from census_data;"""

result = pd.read_sql_query(query , conn)
print(result)
```

```python
query = """
SELECT community_area_name, community_area_number, per_capita_income
FROM census_data
WHERE per_capita_income < 11000;
"""       

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
#List all case numbers for crimes involving minors?(children are not considered minors for the purposes of crime analysis) 

query = """ select case_number from crime_data 
where description like '%MINOR%'
or description not like '%CHILD%' """

result = pd.read_sql_query(query , conn)

print(result.to_string(index=False))
```
